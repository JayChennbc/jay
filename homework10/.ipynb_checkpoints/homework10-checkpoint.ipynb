```python
for col in df:
    if col != "Planet" and col != "Ring System?" and col != "Global Magnetic Field?":
        df[col] = pd.to_numeric(df[col], errors = "coerce")      #isolates the columns we want to change
for col in df:
    print(df[col].apply(type))           #prints diff dtypes
```

```python
#The three different dtypes now in df are strings, floats, and integers
# 'str' = string (a word), 'float64' = float (decimals), 'int64' = integer (whole number)
# Only the column name labels are objects because they are not numerical or word data.
```

```python
df
```

```python
def attach_units(col):
    units = 10**6 * u.km
    data = []
    for val in df[col]:
        data.append(val*units)
    df[col] = data
#favorite column "Distance from Sun"
attach_units("Distance from Sun (10^6 km)")
print(df["Distance from Sun (10^6 km)"])
```

```python
data=[]
for val in df['Orbital Period (days)']:
    units = (val/365)*u.year
    data.append(units)
df['Orbital Period (days)'] = data

print(df['Orbital Period (days)'])
```

```python
df.rename(columns = {"Orbital Period (days)": "Orbital Period (years)"}, inplace=True)
print(df.loc[0])
print(df["Orbital Period (years)"])
```

```python
    #Convert Distance Columns (km --> AU)
#AU stands for astronomical unit and is the average distance between Earth and our sun.
print(const.au)
```

```python
#columns to convert to AU: Diameter (km), Distance from Sun (10^6 km), Perihelion (10^6 km), Aphelion (10^6 km), Semi-Major Axis (10^6 km)
```

```python
kmcolumns = ['Diameter (km)', 'Distance from Sun (10^6 km)', 'Perihelion (10^6 km)', 'Aphelion (10^6 km)', 'Semi-Major Axis (10^6 km)']
for col in kmcolumns:
    newdata = []
    if col != "Diameter (km)":
        for val in df[col]:
            units = (val*10**6/(149597870.70000002))*u.AU      #10^6 km to AU
            newdata.append(units)
        df[col] = newdata
    else:
        for val in df[col]:
            units = (val/(149597870.70000002))*u.AU      #km to AU
            newdata.append(units)
        df[col] = newdata
df
```

```python
df.rename(columns = {"Diameter (km)": "Diameter (AU)"}, inplace=True)
df.rename(columns = {"Distance from Sun (10^6 km)": "Distance from Sun (AU)"}, inplace=True)
df.rename(columns = {"Perihelion (10^6 km)": "Perihelion (AU)"}, inplace=True)
df.rename(columns = {"Aphelion (10^6 km)": "Aphelion (AU)"}, inplace=True)
df.rename(columns = {"Semi-Major Axis (10^6 km)": "Semi-Major Axis (AU)"}, inplace=True)
```

```python
print(df["Diameter (AU)"][4])
print(df["Distance from Sun (AU)"][4])
print(df["Perihelion (AU)"][4])
print(df["Aphelion (AU)"][4])
print(df["Semi-Major Axis (AU)"][4])        #converted values of MARS 
```
