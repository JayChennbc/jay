```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#2 --- Matplotlib ---
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return y

x=np.linspace(0, 2*np.pi, 1000)
A = [0.5, 1, 1.5, 2, 2.5]    #amplitude
w = [1, 2, 3, 4, 5]      #frequency
# Aw = list(zip(A, w))

fig, ax=plt.subplots(3,1, figsize=(25,5))
for i in A:
    for j in w:
        y1 = make_sine_wave(x, i*2, j)
        ax[0].plot(x, y1)
        ax[0].set_title("Vary Amplitude (A)")
        ax[0].set_xlabel("x")
        ax[0].set_ylabel("y1")
for i in A:
    for j in w:
        y2 = make_sine_wave(x, i, j*5)
        ax[1].plot(x, y2)
        ax[1].set_title("Vary Frequency (w)")
        ax[1].set_xlabel("x")
        ax[1].set_ylabel("y2")
for i in A:
    for j in w:
        y3 = make_sine_wave(x, i*5, j*3)
        ax[2].plot(x, y3)
        ax[2].set_title("Vary A and w")
        ax[2].set_xlabel("x")
        ax[2].set_ylabel("y3")
```
