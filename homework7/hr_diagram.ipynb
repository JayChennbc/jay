```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#3.1 --- Importing Data with Numpy ---
star_data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
print(star_data)
```

```python
#3.2 --- Analyzing Imported Data ---
print(star_data.shape)   #2642 rows
magnitude = star_data[:,0]
print(magnitude)
temperature = star_data[:,1]
print(temperature)
print("magnitude min=", np.min(magnitude), "star index=", np.argmin(magnitude))
print("magnitude max=", np.max(magnitude), "star index=", np.argmax(magnitude))
print("temperature min=", np.min(temperature), "star index=", np.argmin(temperature))
print("temperature max=", np.max(temperature), "star index=", np.argmax(temperature))
print("brightest star index=", np.argmin(magnitude), "dimmest star index=", np.argmax(magnitude))
```

```python
#3.3 --- Plot an HR Diagram ---
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
plt.figure(figsize = (8,6))
log_data = plt.scatter(log_temperature, log_magnitude, label = "Other Stars")
print(log_data)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
print(star_data[1988])
print(star_data[264])
print(star_data[2613])
print(star_data[1923])
plt.scatter(np.log(temperature[1988]), np.log(magnitude[1988]), color = "black", s = 90, marker = "*", label = "Brightest Star")
plt.scatter(np.log(temperature[264]), np.log(magnitude[264]), color = "purple", s = 90, marker = "*", label = "Dimmest Star")
plt.scatter(np.log(temperature[2613]), np.log(magnitude[2613]), color = "green", s = 90, marker = "*", label = "Hottest Star")
plt.scatter(np.log(temperature[1923]), np.log(magnitude[1923]), color = "orange", s = 90, marker = "*", label = "Coldest Star")
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
```
